```python
import pandas as pd
df = pd.read_csv('student_performance_dataset.csv')
```

```python
df
```

```python
from scipy import stats
stat, p_valor_shapiro_est = stats.shapiro(df['study_time_hours'])
print(p_valor_shapiro_est)
if p_valor_shapiro_est > 0.05:
    print("Los datos se distribuyen siguiendo la distribución normal.")
else:
    print("Los datos no siguen la distrubución normal.")
```

```python
stat, p_valor_shapiro_dor = stats.shapiro(df['sleep_hours'])
print(p_valor_shapiro_dor)
if p_valor_shapiro_dor > 0.05:
    print("Los datos se distribuyen siguiendo la distribución normal.")
else:
    print("Los datos no siguen la distrubución normal.")
```

```python
from scipy.stats import spearmanr
corr_estudio, p_valor_est = spearmanr(df['study_time_hours'], df['final_exam_score'])
corr_sueño, p_valor_dor = spearmanr(df['sleep_hours'], df['final_exam_score'])

print(f'Horas de estudio VS Nota final: r = {corr_estudio:.2f} (p-valor = {p_valor_est:.2e})')
print(f'Horas de sueño VS Nota final: r= {corr_sueño:.2f} (p-valor = {p_valor_dor:.2e})')
```

```python
matriz_correlacion = df[['study_time_hours', 'sleep_hours', 'final_exam_score']].corr(method='spearman')
print(matriz_correlacion)
```

```python
import statsmodels.api as sm
X = df[['study_time_hours', 'sleep_hours']]
y = df['final_exam_score']

X = sm.add_constant(X)

modelo = sm.OLS(y, X).fit()
print(modelo.summary())
```
